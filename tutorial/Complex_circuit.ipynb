## ANALYZE COMPLEX CIRCUIT

Here we want to test the stability of QuLTRA by analyzing complex circuits with a large number of nodes. The idea is to analyze a chain of six qubits capacitively coupled to each other, with each qubit coupled to its own $\lambda/4$ resonator.

```python
#qubits parameters
C=90e-15 #F
J1=16e-9 #H
J2=14e-9
J3=12e-9
J4=11e-9
J5=10e-9
J6=9e-9

print("The frequency of the first qubit is ", 1/np.sqrt(C*J1)/2/np.pi/1e9, "GHz")
print("The frequency of the second qubit is ", 1/np.sqrt(C*J2)/2/np.pi/1e9, "GHz")
print("The frequency of the third qubit is ", 1/np.sqrt(C*J3)/2/np.pi/1e9, "GHz")
print("The frequency of the fourth qubit is ", 1/np.sqrt(C*J4)/2/np.pi/1e9, "GHz")
print("The frequency of the fifth qubit is ", 1/np.sqrt(C*J5)/2/np.pi/1e9, "GHz")
print("The frequency of the sixth qubit is ", 1/np.sqrt(C*J6)/2/np.pi/1e9, "GHz")
```

```python
#length of the cpw lambda/4 redaout resonators


l1=4.8e-3 #6.15GHz
l2=4.6e-3 #6.4GHz
l3=4.4e-3 #6.7GHz
l4=4.2e-3 #7GHz
l5=4.0e-3 #7.4GHz
l6=3.8e-3 #7.8GHz

netlist=[qu.CPW(0,1,l6)]
circuit=qu.QCircuit(netlist,1,8)
circuit.show_modes()

#qubit-qubit coupling capacitance
Cqq=6e-15 #F

#qubit-resonator coupling capacitance
Cqr=12e-15 #F
```

```python
#total circuit netlist (no feedline)

netlist=[qu.C(0,1,C),qu.J(0,1,J1),qu.C(1,2,Cqq),qu.C(2,0,C),qu.J(0,2,J2),qu.C(2,3,Cqq),qu.C(0,3,C),qu.J(0,3,J3),qu.C(3,4,Cqq),qu.C(0,4,C),qu.J(0,4,J4),
         qu.C(4,5,Cqq),qu.C(0,5,C),qu.J(0,5,J5),qu.C(5,6,Cqq),qu.C(0,6,C),qu.J(0,6,J6),
         qu.C(1,7,Cqr),qu.CPW(0,7,l1),qu.C(2,8,Cqr),qu.CPW(0,8,l2),qu.C(3,9,Cqr),qu.CPW(0,9,l3),qu.C(4,10,Cqr),qu.CPW(0,10,l4),qu.C(5,11,Cqr),qu.CPW(0,11,l5),qu.C(6,12,Cqr),qu.CPW(0,12,l6)]

complex_circuit=qu.QCircuit(netlist,1,10)
complex_circuit.show_all()
```

```python
#adding feedline coupling
Crf=20e-15 #F

feedline_coupling=[qu.C(7,13,Crf),qu.C(8,14,Crf),qu.C(9,15,Crf),qu.C(10,16,Crf),qu.C(11,17,Crf),qu.C(12,18,Crf),
                   qu.CPW(13,14,0.2e-3),qu.CPW(14,15,0.2e-3),qu.CPW(15,16,0.2e-3),qu.CPW(16,17,0.2e-3),qu.CPW(17,18,0.2e-3),
                   qu.CPW(13,19,0.2e-3),qu.CPW(18,20,0.2e-3),qu.R(0,19,50),qu.R(20,0,50)]

total_netlist=netlist+feedline_coupling

dissipative_circuit=qu.QCircuit(total_netlist,1,10)
dissipative_circuit.show_all()
```
